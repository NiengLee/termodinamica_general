<h1 style="text-align:center;">Capítulo 11 - Mezcla no reactivas de gases ideales</h1>

<p style="text-align:center;">En este capitulo se ralizan análisis de mezcla de gases y como es el aporte parcial de cada componente en propiedades de mezcla. Es una etapa introductoria a análisis de sistemas termódinamicos multicomponentes y multifasicos.</p>

# Análisis gravimétrico : 
Es el análisis referido a la masa de una mezcla (no por moles).

Para una mezcla no reactiva de gases, tenemos masta total $m_T$ compuesta de k componentes:

$$
m_{\text{total}} = \Sigma_{i=1}^k m_i 
$$

Fracción masa del componente i (dentro del intervalo de 0 a 1):

$$
mf_i = \frac{m_i}{m_{\text{total}}}\\[10pt]
\Sigma mf_i = 1
$$

# Análisis molar : 
Es el análiss de una mezcla de gases por cantidad de moles, y de manera análoga:

$$
n_T = \Sigma_{i=1}^k n_i 
$$

Fracción molar del componente i (dentro del intervalo de 0 a 1):

$$
y_i = \frac{n_i}{n_{\text{total}}}\\[10pt]
\Sigma y_i = 1
$$

Recuerdese que peso molecular $\overline{PM}$ y :

$$
\overline{PM}_i = \frac{m_i}{n_i} \rightarrow m_i = \overline{PM}_i n_i
$$

## Masa molar aparente (masa molar media):
$$
\overline{PM}_{\text{total}} = \frac{m_{\text{total}}}{n_{\text{total}}} = \frac{\Sigma m_{\text{i}}}{n_{\text{total}}} = \Sigma y_i\overline{PM}_{i} 
$$

# Relaciones Pvt para mezclas de gases ideales
Estos modelos aplican para mezcla de gases ideales y reales

## Modelo de Dalton

$$
P_{\text{total}} = \Sigma_{\text{i=1}}^k p_i
$$

Ahora por medio de gas ideal:

$$
\begin{aligned}
P_{\text{total}} &= \frac{n_{\text{total}}RT}{V}\\[10pt]
p_i &= \frac{n_{\text{i}}RT}{V}
\end{aligned}
$$

por tanto

$$
\begin{aligned}
\frac{p_i}{P_{\text{total}}} &= \frac{\frac{n_{\text{i}}RT}{V}}{\frac{n_{\text{total}}RT}{V}} = \frac{n_i}{n_{\text{total}}} = y_i
\end{aligned}
$$

finalmente la presión parcial se puede calcular como $p_i = y_iP_{\text{total}}$

## Modelo de Amagat

$$
V_{\text{total}} = \Sigma_{\text{i=1}}^k V_i
$$

Ahora por medio de gas ideal:

$$
\begin{aligned}
V_{\text{total}} &= \frac{n_{\text{total}}RT}{P}\\[10pt]
V_i &= \frac{n_{\text{i}}RT}{P}
\end{aligned}
$$

por tanto

$$
\begin{aligned}
\frac{V_i}{V_{\text{total}}} &= \frac{\frac{n_{\text{i}}RT}{P}}{\frac{n_{\text{total}}RT}{P}} = \frac{n_i}{n_{\text{total}}} = y_i
\end{aligned}
$$

finalmente la presión parcial se puede calcular como $V_i = y_iV_{\text{total}}$ y para nuestro caso especial de gases ideales, $\frac{p_i}{P_t} = \frac{V_i}{V_t} = y_i$

# Energía interna, Entalpía, Entropía de mezclas de gases ideales

## Ley de Gibss-Dalton

$$
\begin{aligned}
U_{\text{total}} &= \Sigma_{i=1}^k U_i\\[10pt]
U_{\text{total}} &= \Sigma_{i=1}^k n_i\overline{u}_i = \Sigma_{i=1}^k m_i u_i\\[10pt]
\overline{u}_{\text{total}} &= \Sigma_{i=1}^k y_i\overline{u}_i\\[10pt]
\end{aligned}
$$

por tanto la variación de energía interna puede calcularse de la siguiente manera:

$$
\begin{aligned}
\Delta U_{\text{total}} &= \Sigma_{i=1}^k m_i \Delta u_i\\[10pt]
\Delta U_{\text{total}} &= \Sigma_{i=1}^k n_i \Delta\overline{u}_i\\[10pt]
\Delta \overline{u}_{\text{total}} &= \Sigma_{i=1}^k y_i\Delta \overline{u}_i\\[10pt]
\end{aligned}
$$

### de manera análoga a las entalpías : 

$$
\begin{aligned}
H_{\text{total}} &= \Sigma_{i=1}^k H_i\\[10pt]
H_{\text{total}} &= \Sigma_{i=1}^k n_i\overline{h}_i = \Sigma_{i=1}^k m_i h_i\\[10pt]
\overline{h}_{\text{total}} &= \Sigma_{i=1}^k y_i\overline{h}_i\\[10pt]
\end{aligned}
$$

por tanto la variación de entalpía puede calcularse de la siguiente manera:

$$
\begin{aligned}
\Delta H_{\text{total}} &= \Sigma_{i=1}^k m_i \Delta h_i\\[10pt]
\Delta H_{\text{total}} &= \Sigma_{i=1}^k n_i \Delta\overline{h}_i\\[10pt]
\Delta \overline{h}_{\text{total}} &= \Sigma_{i=1}^k y_i\Delta \overline{h}_i\\[10pt]
\end{aligned}
$$

### Expresiones en terminos de capacidades caloríficas :

$$
\begin{aligned}
\Delta u_{\text{total}} &= C_{\text{v,total}}\Delta T\\[10pt]
\Delta h_{\text{total}} &= C_{\text{p,total}}\Delta T\\[10pt]
C_{\text{v,total}} &= \Sigma y_iC_{\text{v, i}}\\[10pt]
C_{\text{p,total}} &= \Sigma y_iC_{\text{p, i}}\\[10pt]
\end{aligned}
$$

### de manera análoga a las entropías : 

$$
\begin{aligned}
S_{\text{total}} &= \Sigma_{i=1}^k S_i\\[10pt]
S_{\text{total}} &= \Sigma_{i=1}^k n_i\overline{s}_i = \Sigma_{i=1}^k m_i s_i\\[10pt]
\overline{s}_{\text{total}} &= \Sigma_{i=1}^k y_i\overline{s}_i\\[10pt]
\end{aligned}
$$

por tanto la variación de entropía puede calcularse de la siguiente manera:

$$
\begin{aligned}
\Delta S_{\text{total}} &= \Sigma_{i=1}^k m_i \Delta s_i\\[10pt]
\Delta S_{\text{total}} &= \Sigma_{i=1}^k n_i \Delta\overline{s}_i\\[10pt]
\Delta \overline{s}_{\text{total}} &= \Sigma_{i=1}^k y_i\Delta \overline{s}_i\\[10pt]
\end{aligned}
$$

#   Cálculo de entropía específica por componente en una mezcla:

Recuerdese que, para gas ideal:

$$
\mathrm{d}s = C_p \frac{\mathrm{d}T}{T} - R \frac{\mathrm{d}p}{p}
$$

en este caso la presión se deonta con p minúscula porque la presión de cualquier componente a una temperatura T y volumen V, se mide por componente (presión parcial), y no la presión total del sistema P. Así:

$$
\begin{aligned}
\Delta s_i &= c_{\text{p,i}}\mathrm{ln}\frac{T_2}{T_1} - R\mathrm{ln}\frac{p_{\text{i,2}}}{p_{\text{i,1}}}  \\[10pt]
\Delta \overline{s}_i &= \overline{c}_{\text{p,i}}\mathrm{ln}\frac{T_2}{T_1} - R\mathrm{ln}\frac{p_{\text{i,2}}}{p_{\text{i,1}}}  \\[10pt]
{c}_{\text{p,i}}\mathrm{ln}\frac{T_2}{T_1} = {s}_{\text{i,2}}^0 - {s}_{\text{i,1}}^0 \: &; \: \overline{c}_{\text{p,i}}\mathrm{ln}\frac{T_2}{T_1} = \overline{s}_{\text{i,2}}^0 - \overline{s}_{\text{i,1}}^0
\end{aligned}
$$

Para procesos isoentrópicos, como se ha visto en capítulos anteriores:

$$
\frac{T_2}{T_1} = \left(\frac{P_2}{P_1}\right)^{\frac{\gamma_m - 1}{\gamma_m}}
$$

con gamma de mezcla : $\gamma_m = \frac{C_{\text{p,m}}}{C_{\text{v,m}}}$

# Mezcla de gas ideal y vapor

Son aquellas mezclas en fase gaseosa, en las que en su aumento de presión, de manera que la presión de un componente pueda ser superior a la presión de saturación respectiva a la temperatura en la que se encuentra el sistema, y conlleve a la condensación del vapor dando lugar a un cambio de fase.

De igual manera con la disminución de la temperatura, si llega a estar por debajo de la temperatura de saturación respectiva a la presión a la que se encuentra un componente, este empieza a condensar.

El ejemplo común en la que se da este fenomeno es el aire atmosférico, que normalmente tiene vapor de agua en su composición

## Aire atmosférico

Tambien considerado como aire húmedo, y el estudio energético sobre la mezcla aire-vapor se reconoce como **PSICROMETRÍA**

- Temperatura de bulbo seco = temperatura medida de manera convencional y normal con un dispositvo de medicíon de temperatura (termómetro al aire libre)
- Temperatura de rocío = punto de saturación donde el vapor empieza a condensar (primera gota de líquido)
- Aire saturado = mezcla de aire vapor de agua que ya esta saturada de vapor de agua
- Humerdad relativa = presión parcial de vapor en la mezcla / presión de saturación del vapor a la temperatura expuesta $\phi = \frac{p_v}{p_g} = \frac{v_v}{v_g}$
- Relación de humedad = tambien conocida como humedad específica y es la proporción de vapor de agua con aire en la mezcla $\omega = \frac{m_v}{m_a} = \frac{\overline{PM}_vp_v}{\overline{PM}_ap_a} $
- Temperatura saturación adiabática = la temperatura de saturación del agua líquida con la que tiene contacto
- Temperatura de bulbo húmedo = es la temperatura medida en el agua en estado líquido en exposición a la evaporación; es la temperatura mínima que puede alcanzar un sistema aire–agua por enfriamiento evaporativo a presión constante

# Relación de humedad exclusiva para mezcla de gas ideal y vapor:
$$
\begin{aligned}
\omega = \frac{\overline{PM}_vp_v}{\overline{PM}_ap_a} &\rightarrow \frac{\overline{PM}_v = 18.02}{\overline{PM}_a =28.97} = 0.622\\[10pt]
P_{\text{total}} &= p_{\text{vapor}} + p_{\text{aire}}\\[10pt]
\omega &= 0.622\frac{p_v}{P_{\text{total}} - p_v}\\[10pt]
\phi &= \frac{p_v}{p_g}\\[10pt]
\checkmark \omega &= 0.622\frac{\phi p_g}{P_{\text{total}} - \phi p_g}\\[10pt]
\checkmark \phi &= \frac{\omega p_a}{0.622p_g}\\[10pt]
\end{aligned}
$$

Se puede usar ecuación de Antoine específica para agua, que determina la presión de saturación a un valor de temperatura es:

$$
\mathrm{ln}p_g = 12.1929 - \frac{4109.1}{T - 35.50}
$$

- $p_g$ [=] bar
- $T$ [=] K

### Entalpía de mezcla aire seco - vapor de agua (aire humedo)

$$
\begin{aligned}
h_{\text{mezcla}} &= h_{\text{aire seco}} + \omega h_{\text{vapor de agua}}\\[10pt]
H_{\text{mezcla}} &= m_{\text{total de mezcla}}.h_{\text{mezcla}}\\[10pt]
\end{aligned}
$$

Normalmente, $h_{\text{vapor de agua}} = h_{\text{g}}$ donde g es el estado de vapor sobrecalentado

Recuerdese que:

$$
c_{\text{p,i}} = \frac{h_i - h_{\text{i,referencia}}}{T - T_{\text{i,referencia}}}
$$

- donde i puede ser a, v (o también g), y f (estado de líquido saturado)
- El estado de referencia normalmente es 0°C

Ahora para las condiciones de temperatura entre 0°C y 40°C, las siguientes expresiones se desempeñan bien :

$$
\begin{aligned}
c_{\text{p,a}} &= 1.005\\[10pt]
c_{\text{p,v}} &= 1.82\\[10pt]
h_{\text{g, referencia}} &= 2501.7\\[10pt]
h_{\text{f, referencia}} &= 4.190\\[10pt]
h_{\text{mezcla}} &= 1.005T_{\text{bulbo seco}} + \omega [2501.7 + 1.82T_{\text{bulbo seco}}]\\[10pt]
h_{\text{f}} &= 4.190T_{\text{bulbo seco}}\\[10pt]
\end{aligned}
$$

### Volumen específico para aire humedo

Se asume que el volumen ocupado por la mezcla, es aproximadamente igual al volumen ocupado por el aire seco:

$$
v_m = \frac{V_m}{m_a} = \frac{V_a}{m_a} = v_a = \frac{RT}{p_a} = \frac{RT}{P_{\text{total}}-p_v}
$$

para efectos prácticos, el peso molecular de esta mezcla aire-vapor puede aproximarse al peso molecular del aire seco.

# Temperatura de saturación adiabática y temperatura de bulbo húmedo (exclusiva para mezcla de gas ideal y vapor) : 

- La temperatura de saturación adiabática es la temperatura en el cual $\phi = \frac{p_v}{p_g} = 1$, 
- la temperatura de saturación adiabática de una mezcla es la temperatura de saturación del agua líquida con la que tiene contacto



Si se resolviera el balance de energía para el sistema anterior partiendo de:

$$
0 = \dot{m}_{\text{a1}}{h}_{\text{m,1}} - \dot{m}_{\text{a2}}{h}_{\text{m,2}} + \dot{m}_{\text{3}}{h}_{\text{3}}
$$

se llega a :

$$
\omega_{1} = \frac{(h_{\text{a,2}} - h_{\text{a,1}}) + \omega_{2}(h_{\text{v,2}} - h_{\text{3}})}{h_{\text{v,1}} - h_{\text{3}}}
$$

ahora, sabiendo que:

- $(h_{\text{a,2}} - h_{\text{a,1}}) = c_{\text{p,a}}(T_{\text{sa,1}} - T_{\text{1}})$
- la relación de humedad $\omega_{2} = \omega_{sat,2}$ ya que sale a condiciones de saturación de vapor
- $(h_{\text{v,2}} - h_{\text{3}}) = h_{\text{fg,2}}$
- $h3 = h_{\text{2,f}}$
- $h_{\text{v,1}} = h_{\text{g,1}}$ ya que esta vapor esta próximo a cambiar de fase

$$
\omega_{1} = \frac{c_{\text{p,a}}(T_{\text{sa,1}} - T_{\text{1}}) + \omega_{sat,2}(h_{\text{fg,2}})}{h_{\text{g,1}} - h_{\text{f,2}}}
$$

- $T_{\text{sa,1}} = T_{\text{bulbo humedo,1}}$

# Aplicación a procesos industriales

### Relación básica de partida:



Balance de masa y energía:

$$
\begin{aligned}
0 &= \dot{m}_1 + \dot{m}_2 + \dot{m}_3 - \dot{m}_4\\[10pt]
0 &= \dot{Q} + (\dot{m}_{\text{a1}}h_{\text{a1}} + \dot{m}_{\text{v1}}h_{\text{v1}}) + (\dot{m}_{\text{a2}}h_{\text{a2}} + \dot{m}_{\text{v2}}h_{\text{v2}}) + \dot{m}_3h_{\text{3}} - (\dot{m}_{\text{a4}}h_{\text{a4}} + \dot{m}_{\text{v4}}h_{\text{v4}})\\[10pt]
0 &= \dot{Q} + \dot{m}_{\text{a1}}(h_{\text{a1}} - h_{\text{a4}}) + \dot{m}_{\text{a2}}(h_{\text{a2}} - h_{\text{a4}}) + \dot{m}_{\text{v1}}(h_{\text{v1}} - h_{\text{v4}}) + \dot{m}_{\text{v2}}(h_{\text{v2}} - h_{\text{v4}}) + \dot{m}_{\text{3}}(h_{\text{3}} - h_{\text{v4}})\\[10pt]
\end{aligned}
$$

### Procesos industriales
- Calentamiento / enfriamiento a relación de humedad constante.
- Deshumidificación con calentamiento.
- Humidificación.
- Enfriamiento por evaporación.
- Mezcla de dos corrientes de aire.
